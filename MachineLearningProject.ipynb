```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "ML_Dataset.csv",
    sep=";",
    quotechar='"',
    quoting=3
)
```

```python
df = df.map(lambda x: x.replace('"', '') if isinstance(x, str) else x)
df.columns = df.columns.str.replace('"', '', regex=False)
df.head()
```

```python
# ============================================================
# EDA: Steps 1–8
# ============================================================

# ------------------------------------------------------------
# 1. UNDERSTAND THE DATASET
# ------------------------------------------------------------

print("=" * 60)
print("1. DATASET OVERVIEW")
print("=" * 60)

print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)


# ------------------------------------------------------------
# 2. DATA QUALITY CHECK
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("2. DATA QUALITY")
print("=" * 60)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nUnique values in categorical columns:")
for col in df.select_dtypes(include="object").columns:
    print(f"\n{col}:")
    print(df[col].unique())


# ------------------------------------------------------------
# 3. UNIVARIATE ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("3. UNIVARIATE ANALYSIS")
print("=" * 60)

# Numerical variables
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

print("\nNumerical variables:")
print(numerical_cols)

print("\nSummary statistics:")
display(df[numerical_cols].describe())

# Categorical variables
categorical_cols = df.select_dtypes(include="object").columns.tolist()

print("\nCategorical variables:")
print(categorical_cols)

for col in categorical_cols:
    print(f"\n{col} distribution:")
    display(df[col].value_counts())


# ------------------------------------------------------------
# 4. BIVARIATE ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("4. BIVARIATE ANALYSIS")
print("=" * 60)

# Target variable
target = "y"

print("\nResponse rate:")
print(df[target].value_counts(normalize=True) * 100)

# Response rate by categorical variables
for col in categorical_cols:
    if col != target:
        print(f"\nResponse rate by {col}:")
        response_rate = (
            df.groupby(col)[target]
            .apply(lambda x: (x == "yes").mean() * 100)
            .sort_values(ascending=False)
        )
        display(response_rate)


# Numerical variables vs target
print("\nNumerical variables by response:")
display(
    df.groupby(target)[numerical_cols]
      .mean()
      .T
)


# ------------------------------------------------------------
# 5. MULTIVARIATE ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("5. MULTIVARIATE ANALYSIS")
print("=" * 60)

# Correlation matrix for numerical variables
correlation = df[numerical_cols].corr()

plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Matrix - Numerical Variables")
plt.show()


# ------------------------------------------------------------
# 6. OUTLIER ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("6. OUTLIER ANALYSIS")
print("=" * 60)

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {outliers} potential outliers")


# ------------------------------------------------------------
# 7. TARGET / RESPONSE ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("7. TARGET ANALYSIS")
print("=" * 60)

target_counts = df[target].value_counts()

print("\nTarget counts:")
print(target_counts)

print("\nTarget percentages:")
print((target_counts / len(df) * 100).round(2))

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x=target)
plt.title("Target Distribution")
plt.xlabel("Subscribed to Term Deposit")
plt.ylabel("Number of Customers")
plt.show()


# ------------------------------------------------------------
# 8. FEATURE SUITABILITY / DATA LEAKAGE CHECK
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("8. FEATURE & DATA LEAKAGE CHECK")
print("=" * 60)

print("""
Feature suitability for the response prediction problem:

TARGET:
    y → whether the customer subscribed to the term deposit

POTENTIAL LEAKAGE:
    duration → duration of the current call.
    This information is only known after/during the current
    contact and therefore should NOT be used when predicting
    whether a customer should be contacted.

DECISION:
    Keep 'duration' in the raw dataset for EDA/documentation,
    but exclude it from the prediction modelling dataset.

Other variables will be evaluated based on whether they would
be available at the time the marketing decision is made.
""")
```

```python
# Numerical columns
numeric_cols = [
    "age",
    "balance",
    "day",
    "duration",
    "campaign",
    "pdays",
    "previous"
]

# Convert numerical columns to numeric datatype
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Categorical columns
categorical_cols = [
    "job",
    "marital",
    "education",
    "default",
    "housing",
    "loan",
    "contact",
    "month",
    "poutcome",
    "y"
]

# Convert categorical columns to category datatype
for col in categorical_cols:
    df[col] = df[col].astype("category")

# Check datatypes
print(df.dtypes)
```

```python
# Response variable for calculations only
df["y_num"] = (df["y"] == "yes").astype(int)


# ============================================================
# 1. AGE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 4))
sns.histplot(df["age"], bins=30, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()


# ============================================================
# 2. BALANCE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 4))
sns.histplot(df["balance"], bins=50, kde=True)
plt.xlim(-1000, 4000)
plt.title("Balance Distribution")
plt.xlabel("Balance")
plt.ylabel("Number of Customers")
plt.show()


# ============================================================
# 3. CAMPAIGN CONTACT DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 4))
sns.histplot(df["campaign"], bins=30, discrete=True)
plt.xlim(0, 15)
plt.title("Distribution of Contacts During Current Campaign")
plt.xlabel("Number of Contacts")
plt.ylabel("Number of Customers")
plt.show()


# ============================================================
# 4. PAIRPLOT — SELECTED NUMERICAL VARIABLES
# ============================================================

pairplot_data = df.sample(2000, random_state=42)

sns.pairplot(
    pairplot_data,
    vars=["age", "balance", "campaign", "previous"],
    hue="y"
)
plt.show()


# ============================================================
# 5. RESPONSE RATE BY AGE GROUP
# ============================================================

df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=[
        "18–25",
        "26–35",
        "36–45",
        "46–55",
        "56–65",
        "65+"
    ]
)

age_response = (
    df.groupby("age_group", observed=False)["y_num"]
      .mean() * 100
)

plt.figure(figsize=(8, 4))
sns.barplot(
    x=age_response.index,
    y=age_response.values
)
plt.title("Response Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Response Rate (%)")
plt.show()


# ============================================================
# 6. RESPONSE RATE BY BALANCE GROUP
# ============================================================

df["balance_group"] = pd.qcut(
    df["balance"],
    q=5,
    duplicates="drop"
)

balance_response = (
    df.groupby("balance_group", observed=False)["y_num"]
      .mean() * 100
)

plt.figure(figsize=(9, 4))
sns.barplot(
    x=balance_response.index.astype(str),
    y=balance_response.values
)
plt.title("Response Rate by Balance Group")
plt.xlabel("Balance Group")
plt.ylabel("Response Rate (%)")
plt.xticks(rotation=30)
plt.show()


# ============================================================
# 7. RESPONSE RATE VS CAMPAIGN CONTACTS
# ============================================================

campaign_response = (
    df.groupby("campaign")["y_num"]
      .mean() * 100
)

# Focus on commonly observed contact counts
campaign_response = campaign_response[campaign_response.index <= 15]

plt.figure(figsize=(8, 4))
sns.lineplot(
    x=campaign_response.index,
    y=campaign_response.values,
    marker="o"
)
plt.title("Response Rate vs Number of Campaign Contacts")
plt.xlabel("Number of Contacts")
plt.ylabel("Response Rate (%)")
plt.grid(True)
plt.show()


# ============================================================
# 8. PREVIOUS CONTACTS × DAYS SINCE PREVIOUS CONTACT
# ============================================================

# Limit to commonly occurring values for readability
pivot = pd.pivot_table(
    df,
    values="y_num",
    index="previous",
    columns="pdays",
    aggfunc="mean"
)

pivot = pivot.loc[pivot.index <= 5]
pivot = pivot.iloc[:, :10]

plt.figure(figsize=(10, 5))
sns.heatmap(
    pivot * 100,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu"
)
plt.title("Response Rate: Previous Contacts × Days Since Previous Contact")
plt.xlabel("Days Since Previous Contact (pdays)")
plt.ylabel("Number of Previous Contacts")
plt.show()


# ============================================================
# 9. RESPONSE RATE BY PREVIOUS CAMPAIGN OUTCOME
# ============================================================

poutcome_response = (
    df.groupby("poutcome", observed=False)["y_num"]
      .mean() * 100
)

plt.figure(figsize=(8, 4))
sns.barplot(
    x=poutcome_response.index,
    y=poutcome_response.values
)
plt.title("Response Rate by Previous Campaign Outcome")
plt.xlabel("Previous Campaign Outcome")
plt.ylabel("Response Rate (%)")
plt.show()
```

```python
# Copy the cleaned dataset
mdf = df.copy()

# Remove columns that should not be used for prediction
mdf = mdf.drop(
    columns=["duration", "y_num", "age_group", "balance_group"],
    errors="ignore"
)

# Convert target into 0 and 1
y = mdf["y"].map({"no": 0, "yes": 1})

# Separate predictors
X = mdf.drop(columns=["y"])

# Identify categorical and numerical columns
categorical_cols = X.select_dtypes(
    include=["object", "category"]
).columns

numerical_cols = X.select_dtypes(
    exclude=["object", "category"]
).columns

# One-hot encode all categorical variables
X_transformed = pd.get_dummies(
    X,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

# Display the transformed dataset
print("Original shape:", X.shape)
print("Transformed shape:", X_transformed.shape)

X_transformed.head()
```

```python
X_transformed.shape
```

```python
from sklearn.preprocessing import StandardScaler
# One common encoded dataset
X_transformed = X_transformed.astype("float32")

# Scaled version for Logistic Regression, K-Means and PCA
scaler = StandardScaler()

X_scaled = pd.DataFrame(
    scaler.fit_transform(X_transformed),
    columns=X_transformed.columns,
    index=X_transformed.index
)

print("Encoded dataset shape:", X_transformed.shape)
print("Scaled dataset shape:", X_scaled.shape)
```

```python
# Logistic Regression
X_logistic = X_scaled

# Decision Tree
X_tree = X_transformed

# Random Forest
X_rf = X_transformed

# XGBoost
X_xgb = X_transformed

# K-Means
X_kmeans = X_scaled

# PCA
X_pca = X_scaled
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_transformed,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Scaled versions for Logistic Regression
X_train_scaled = X_scaled.loc[X_train.index]
X_test_scaled = X_scaled.loc[X_test.index]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train_scaled, y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

tree_model.fit(X_train, y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=4,
    random_state=42,
    eval_metric="logloss"
)

xgb_model.fit(X_train, y_train)
```

```python
models = {
    "Logistic Regression": (logistic_model, X_test_scaled),
    "Decision Tree": (tree_model, X_test),
    "Random Forest": (rf_model, X_test),
    "XGBoost": (xgb_model, X_test)
}

predictions = {}

for name, (model, features) in models.items():
    predictions[name] = model.predict(features)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

results = []

for name, (model, features) in models.items():
    y_pred = model.predict(features)
    y_prob = model.predict_proba(features)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)
results_df.sort_values("ROC-AUC", ascending=False)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

for name, (model, features) in models.items():
    y_pred = model.predict(features)

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        y_pred,
        display_labels=["No", "Yes"]
    )

    plt.title(f"Confusion Matrix - {name}")
    plt.show()
```

```python
from sklearn.metrics import roc_curve, auc

plt.figure(figsize=(8, 6))

for name, (model, features) in models.items():
    y_prob = model.predict_proba(features)[:, 1]
    fpr, tpr, _ = roc_curve(y_test, y_prob)

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC = {auc(fpr, tpr):.3f})"
    )

plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()
plt.show()
```

```python
balanced_results = []

for name, (model, features) in balanced_models.items():

    y_pred = model.predict(features)
    y_prob = model.predict_proba(features)[:, 1]

    balanced_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test,
            y_pred,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test,
            y_pred,
            zero_division=0
        ),
        "F1 Score": f1_score(
            y_test,
            y_pred,
            zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

balanced_results_df = pd.DataFrame(balanced_results)

balanced_results_df = balanced_results_df.sort_values(
    "ROC-AUC",
    ascending=False
)

balanced_results_df
```

```python
# Use the same transformed features, but exclude the target
X_segmentation = X_transformed.copy()

# Scale the features because K-Means is distance-based
X_segmentation_scaled = X_scaled.copy()

print(X_segmentation_scaled.shape)
```

```python
from sklearn.cluster import KMeans
inertia = []
k_values = range(2, 11)

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_segmentation_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(k_values, inertia, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method for Selecting Number of Clusters")
plt.show()
```

```python
X_segmentation = X_transformed.drop(
    columns=["y_num", "y", "duration", "day"],
    errors="ignore"
)

from sklearn.preprocessing import StandardScaler

scaler_seg = StandardScaler()
X_segmentation_scaled = scaler_seg.fit_transform(X_segmentation)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_segmentation_scaled)

df_segmented = df.copy()
df_segmented["Cluster"] = cluster_labels
```

```python
cluster_summary = df_segmented.groupby("Cluster").mean(
    numeric_only=True
).round(2)

cluster_summary
```

```python
from sklearn.decomposition import PCA
from plotnine import ggplot, aes, geom_point, labs, theme_minimal

# PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_segmentation_scaled)

# Create plotting dataframe
pca_df = pd.DataFrame({
    "PCA1": X_pca[:, 0],
    "PCA2": X_pca[:, 1],
    "Cluster": cluster_labels.astype(str)
})

# Plot
(
    ggplot(pca_df, aes(x="PCA1", y="PCA2", color="Cluster"))
    + geom_point(alpha=0.6)
    + labs(
        title="Customer Segmentation using K-Means and PCA",
        x="PCA 1",
        y="PCA 2"
    )
    + theme_minimal()
)
```
